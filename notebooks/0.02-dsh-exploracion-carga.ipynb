### 1.1 Descarga de los archivos

Las URLs del portal no se pueden construir a partir del año y el mes: cada archivo vive en una carpeta con un identificador distinto (`/site/assets/files/{id}/datos_abiertos_AAAA_MM.csv`). Por eso los enlaces se leen directamente de la página del portal y se filtran por el rango de meses.

Si falta en el portal algún mes del rango que no esté registrado en `MESES_SIN_PUBLICAR`, el notebook se detiene con un error. Así un cambio nuevo del portal no pasa desapercibido en el resto del análisis.

Cada archivo se descarga una sola vez: si ya existe en `data/raw/`, se omite.

```python
def leer_portal(url_portal):
    """Regresa el HTML de la página de datos abiertos."""
    respuesta = requests.get(url_portal, headers=ENCABEZADOS_HTTP, timeout=30)
    respuesta.raise_for_status()
    return respuesta.text


def extraer_enlaces(html, url_base, prefijo):
    """Regresa un DataFrame con archivo, periodo y URL de los CSV cuyo nombre inicia con el prefijo."""
    patron = rf"(?P<archivo>{prefijo}_(?P<anio>\d{{4}})_(?P<mes>\d{{2}})\.csv)"
    enlaces = {}
    for href in re.findall(r"href=['\"]([^'\"]+)['\"]", html):
        coincidencia = re.search(patron, href)
        if coincidencia:
            enlaces[coincidencia["archivo"]] = (pd.Period(f"{coincidencia['anio']}-{coincidencia['mes']}", freq="M"),
                                                urljoin(url_base, href))
    tabla = pd.DataFrame([(archivo, periodo, url) for archivo, (periodo, url) in enlaces.items()],
                         columns=["archivo", "periodo", "url"])
    return tabla.sort_values("periodo").reset_index(drop=True)


def descargar_archivo(url, destino):
    """Descarga el archivo si no existe; regresa True si se descargó en esta ejecución."""
    if destino.exists():
        return False
    respuesta = requests.get(url, headers=ENCABEZADOS_HTTP, timeout=60)
    respuesta.raise_for_status()
    destino.write_bytes(respuesta.content)
    return True
```

### 1.3 Inventario de los archivos de viajes

Antes de unir los archivos se revisa, para cada uno:
- **Codificación de caracteres:** se intenta leer como UTF-8 (detectando si trae BOM) y, si falla, se usa latin-1. UTF-8 tiene reglas estrictas, así que un archivo en latin-1 con acentos o "ñ" no pasa como UTF-8; latin-1 decodifica cualquier byte, por eso es el último recurso.
- **Registros:** número de líneas de datos, sin contar el encabezado.
- **Columnas:** nombres exactos tal como vienen en el archivo.
- **Formatos de fecha:** sobre una muestra de 2000 filas, cada valor con hora se convierte a su "forma", reemplazando cada dígito por `9` (por ejemplo, `2020-01-01 00:03:24` se convierte en `9999-99-99 99:99:99`). Así se detecta cualquier formato distinto sin suponer de antemano cuáles existen. Se investigó la técnica: es el perfilado de patrones (*pattern profiling*), una práctica estándar de data profiling para identificar los formatos presentes en una columna.

El inventario solo registra nombres de columnas y formatos; no muestra valores de las columnas con datos de usuarios.

```python
def detectar_codificacion(contenido):
    """Regresa la codificación con la que se pueden decodificar los bytes del archivo."""
    if contenido.startswith(codecs.BOM_UTF8):
        return "utf-8-sig"
    try:
        contenido.decode("utf-8")
        return "utf-8"
    except UnicodeDecodeError:
        return "latin-1"  # decodifica cualquier byte; se usa como último recurso


def formas_de_fecha(muestra):
    """Regresa las formas (dígitos como 9) de los valores con hora en la muestra."""
    formas = set()
    for columna in muestra.columns:
        valores = muestra[columna].dropna()
        if valores.str.contains(":").any():
            formas.update(valores.str.replace(r"\d", "9", regex=True).unique())
    return sorted(formas)


def inventariar_archivo(ruta, filas_muestra=2000):
    """Regresa codificación, registros, columnas y formatos de fecha de un CSV."""
    contenido = ruta.read_bytes()
    codificacion = detectar_codificacion(contenido)
    lineas = contenido.count(b"\n") + (0 if contenido.endswith(b"\n") else 1)
    muestra = pd.read_csv(ruta, encoding=codificacion, nrows=filas_muestra, dtype=str)
    return {"archivo": ruta.name,
            "tamano_mb": round(len(contenido) / 1024**2, 1),
            "codificacion": codificacion,
            "registros": lineas - 1,
            "n_columnas": muestra.shape[1],
            "columnas": " | ".join(muestra.columns),
            "formatos_fecha": " | ".join(formas_de_fecha(muestra)),
    }
```

```python
# Inventario de cada archivo de viajes del rango
inventario = pd.DataFrame([inventariar_archivo(RUTA_VIAJES_CRUDOS / archivo)
                           for archivo in enlaces_viajes["archivo"]])
inventario.insert(1, "periodo", enlaces_viajes["periodo"].astype(str))

print("Archivos inventariados:", len(inventario))
print("Registros totales:", inventario["registros"].sum())
print(f"Tamaño total: {inventario['tamano_mb'].sum():.1f} MB")
inventario[["archivo", "periodo", "tamano_mb", "codificacion", "registros", "n_columnas"]]
```

```python
# Esquemas distintos: archivos que comparten exactamente los mismos nombres de columnas
esquemas = (inventario.groupby("columnas")
            .agg(archivos=("archivo", "count"),
                 primer_periodo=("periodo", "min"),
                 ultimo_periodo=("periodo", "max"),
                 periodos=("periodo", lambda periodos: ", ".join(periodos)))
            .sort_values("archivos", ascending=False)
            .reset_index()
)
print("Esquemas de columnas distintos:", len(esquemas))
esquemas
```

```python
# Codificación de caracteres por archivo: conteo y porcentaje
conteo_codificacion = pd.concat([inventario["codificacion"].value_counts(),
                                 inventario["codificacion"].value_counts(normalize=True).mul(100).round(2)],
                                axis=1, keys=["count", "percent"],
)
print(conteo_codificacion)
print("-" * 45)

# Archivos que no están en UTF-8 simple
print(inventario.loc[inventario["codificacion"] != "utf-8", ["archivo", "codificacion"]].to_string(index=False))
```

```python
# Formatos de fecha distintos y en qué archivos aparecen
formatos = (inventario.groupby("formatos_fecha")
            .agg(archivos=("archivo", "count"),
                 periodos=("periodo", lambda periodos: ", ".join(periodos)))
            .sort_values("archivos", ascending=False)
            .reset_index()
)
print("Combinaciones de formatos de fecha:", len(formatos))
formatos
```

**Observaciones del inventario de viajes:**
- **Volumen:** 27086557 registros en 81 archivos. El mes con menos viajes es 2020-04 (129276), que coincide con el inicio de la pandemia de COVID-19; el de más viajes es 2026-09 (490056).
- **Columnas:** todos los archivos tienen las mismas 8 columnas en el mismo orden, pero hay 2 esquemas. 79 archivos usan `Año_de_nacimiento` y 2 archivos (2024-02 y 2024-05) traen el nombre corrupto `A}äe_nacimiento`. Si se unieran sin corregirlo, pandas crearía dos columnas distintas y llenaría la otra con nulos; el nombre debe normalizarse antes de unir los archivos.
- **Codificación:** 43 archivos (53.09%) no son UTF-8 válido y se leen como latin-1; los otros 38 (46.91%) son UTF-8. Los archivos latin-1 se concentran entre 2021-06 y 2025-02, con excepciones en UTF-8 dentro de ese periodo (2022-12, 2023-04, 2023-05 y 2024-07) y dos casos aislados después (2025-07 y 2026-01). Ningún archivo trae BOM; esa detección se conserva solo como precaución. Como no sigue una regla por fecha, la codificación se detecta archivo por archivo en lugar de fijar una sola. La detección es confiable porque el encabezado de todos los archivos contiene al menos un carácter no ASCII (la "ñ" de `Año_de_nacimiento` o la "ä" del nombre corrupto).
- **Formatos de fecha:** 78 archivos usan `AAAA-MM-DD hh:mm:ss`. Tres archivos (2023-11, 2024-09 y 2024-10) usan día, mes y año separados por barras, sin segundos y con la hora en uno o dos dígitos (`9:99` y `99:99`). En 2023-11 solo apareció la hora de un dígito, probablemente porque la muestra cubre solo las primeras horas del mes. Por esto, al unir los archivos se usará un formato explícito por archivo en lugar de dejar que pandas lo infiera, ya que con formatos mezclados podría intercambiar día y mes sin avisar. La forma no distingue `DD/MM` de `MM/DD`; el orden se verificará comprobando que el mes de cada fecha coincida con el mes del archivo.
- **Limitación:** los formatos se detectaron en las primeras 2000 filas de cada archivo; la conversión completa de las fechas confirmará que no existen otros.

### 1.5 Lectura y unión de los archivos de viajes

Con lo encontrado en el inventario, cada archivo se lee con:
- **Su codificación detectada** (UTF-8 o latin-1).
- **El nombre corrupto corregido:** `A}äe_nacimiento` pasa a `Año_de_nacimiento`. Después se verifica que todos los archivos tengan exactamente las mismas 8 columnas.
- **Un formato de fecha explícito según su forma**, en lugar de dejar que pandas lo infiera. Los archivos con barras se leen como `DD/MM/AAAA`.
- **Tipos ligeros para reducir memoria:** IDs y año de nacimiento como enteros que aceptan nulos (`Int64`, `Int32`, `Int16`) y `Genero` como `category`.

Los valores se leen primero como texto y luego se convierten. Así se puede contar cuántos no pudieron convertirse, en lugar de que la lectura falle o los pierda sin aviso. También se agrega `periodo_archivo` para conservar el archivo de origen de cada viaje.

Validaciones por archivo:
1. Registros leídos contra los contados en el inventario.
2. Valores que no se pudieron convertir a su tipo.
3. Viajes cuyo inicio no cae en el mes del archivo. Confirma el orden día/mes de los archivos con barras: si fuera `MM/DD`, casi ningún viaje coincidiría con su mes.

El resultado se guarda en **Parquet** en `data/interim/`, con el sufijo `_exploracion` y los nombres de columna originales, para no sobrescribir el Parquet de la ingesta del proyecto (`dataset.py`), que usa nombres normalizados. Es un formato columnar y comprimido que conserva los tipos, por lo que el EDA y los siguientes notebooks lo cargan en segundos sin repetir esta lectura.

```python
# Formato explícito para cada forma de fecha encontrada en el inventario
FORMATOS_FECHA = {"9999-99-99 99:99:99": "%Y-%m-%d %H:%M:%S",
                  "99/99/9999 99:99": "%d/%m/%Y %H:%M",
                  "99/99/9999 9:99": "%d/%m/%Y %H:%M",
}
CORRECCION_COLUMNAS = {"A}äe_nacimiento": "Año_de_nacimiento"}
COLUMNAS_VIAJES = ["Viaje_Id", "Usuario_Id", "Genero", "Año_de_nacimiento",
                   "Inicio_del_viaje", "Fin_del_viaje", "Origen_Id", "Destino_Id"]
TIPOS_ENTEROS = {"Viaje_Id": "Int64", "Usuario_Id": "Int64", "Año_de_nacimiento": "Int16",
                 "Origen_Id": "Int32", "Destino_Id": "Int32"}


def formato_del_archivo(formas):
    """Regresa el único formato de fecha que corresponde a las formas detectadas en un archivo."""
    formatos = {FORMATOS_FECHA[forma] for forma in formas.split(" | ")}
    if len(formatos) != 1:
        raise ValueError(f"El archivo mezcla formatos de fecha: {formas}")
    return formatos.pop()


def leer_viajes(fila):
    """Lee un CSV de viajes con columnas, fechas y tipos normalizados; regresa los viajes y su validación."""
    viajes = pd.read_csv(RUTA_VIAJES_CRUDOS / fila.archivo, encoding=fila.codificacion, dtype=str)
    viajes = viajes.rename(columns=CORRECCION_COLUMNAS)
    if list(viajes.columns) != COLUMNAS_VIAJES:
        raise ValueError(f"{fila.archivo} tiene columnas inesperadas: {list(viajes.columns)}")

    formato = formato_del_archivo(fila.formatos_fecha)
    validacion = {"archivo": fila.archivo, "registros_inventario": fila.registros,
                  "registros_leidos": len(viajes)}
    for columna in COLUMNAS_VIAJES:
        if columna in ("Inicio_del_viaje", "Fin_del_viaje"):
            convertida = pd.to_datetime(viajes[columna], format=formato, errors="coerce")
        elif columna in TIPOS_ENTEROS:
            convertida = pd.to_numeric(viajes[columna], errors="coerce").astype(TIPOS_ENTEROS[columna])
        else:
            continue
        # Valores que venían con dato pero no se pudieron convertir
        validacion[f"no_convertidos_{columna}"] = int((viajes[columna].notna() & convertida.isna()).sum())
        viajes[columna] = convertida

    periodo = pd.Period(fila.periodo, freq="M")
    inicio = viajes["Inicio_del_viaje"]
    en_mes = (inicio.dt.year == periodo.year) & (inicio.dt.month == periodo.month)
    validacion["inicio_fuera_del_mes"] = int((inicio.notna() & ~en_mes).sum())
    viajes["periodo_archivo"] = fila.periodo
    return viajes, validacion
```

```python
# Lectura de cada archivo del inventario y unión en un solo DataFrame
partes, validaciones = [], []
for fila in inventario.itertuples():
    viajes_mes, validacion = leer_viajes(fila)
    partes.append(viajes_mes)
    validaciones.append(validacion)

viajes = pd.concat(partes, ignore_index=True)
del partes  # libera la memoria de las partes antes de continuar

# Se convierten a category después de unir para que compartan las mismas categorías
viajes["Genero"] = viajes["Genero"].astype("category")
viajes["periodo_archivo"] = viajes["periodo_archivo"].astype("category")
validaciones = pd.DataFrame(validaciones).set_index("archivo")

print("Archivos leídos:", len(validaciones))
print("Registros unidos:", len(viajes))
print(f"Memoria del DataFrame: {viajes.memory_usage(deep=True).sum() / 1024**2:.1f} MB")
```

```python
# Validación 1: registros leídos contra el inventario
print("Registros en el inventario:", validaciones["registros_inventario"].sum())
print("Registros leídos          :", validaciones["registros_leidos"].sum())
diferencias = validaciones[validaciones["registros_inventario"] != validaciones["registros_leidos"]]
print("Archivos con diferencias  :", len(diferencias))
print("-" * 45)

# Validación 2: valores que no se pudieron convertir, por columna
columnas_conversion = [columna for columna in validaciones.columns if columna.startswith("no_convertidos_")]
print(validaciones[columnas_conversion].sum().to_string())
print("-" * 45)

# Validación 3: viajes cuyo inicio no cae en el mes del archivo
print("Viajes con inicio fuera del mes del archivo:", validaciones["inicio_fuera_del_mes"].sum())

# Archivos con alguna diferencia en cualquiera de las tres validaciones
con_hallazgos = validaciones[(validaciones["registros_inventario"] != validaciones["registros_leidos"])
                             | (validaciones[columnas_conversion].sum(axis=1) > 0)
                             | (validaciones["inicio_fuera_del_mes"] > 0)]
print("Archivos con hallazgos:", len(con_hallazgos))
con_hallazgos
```

```python
# Guardado del dataset unido en Parquet para el EDA y los siguientes notebooks
# Sufijo _exploracion: no sobrescribe el Parquet de la ingesta del proyecto (dataset.py)
ruta_parquet = RUTA_INTERMEDIOS / f"viajes_{MES_INICIO}_{MES_FIN}_exploracion.parquet"
viajes.to_parquet(ruta_parquet, index=False)

print("Archivo:", ruta_parquet.name)
print(f"Tamaño en Parquet: {ruta_parquet.stat().st_size / 1024**2:.1f} MB")
print(f"Tamaño de los CSV: {inventario['tamano_mb'].sum():.1f} MB")
```

**Observaciones de la lectura y unión:**
- **Registros:** se leyeron 27086557 registros de 81 archivos, exactamente los contados en el inventario; no se perdió ni se duplicó ninguno. Esto también confirma que ningún campo contiene saltos de línea.
- **Conversión de tipos:** ningún valor con dato quedó sin convertir. Los tipos elegidos alcanzan para todos los valores y los formatos explícitos cubren todas las fechas, no solo las 2000 filas de muestra del inventario.
- **Orden día/mes:** ningún viaje inicia fuera del mes de su archivo, lo que confirma que los archivos con barras usan `DD/MM/AAAA`.
- **Tamaño:** el DataFrame ocupa 1265.8 MB en memoria. El Parquet pesa 541.9 MB, cerca de una cuarta parte de los 2155.1 MB de los CSV.
